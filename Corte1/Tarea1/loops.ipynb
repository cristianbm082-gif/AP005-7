Obtiene el factorial de un número entero positivo ingresado por el usuario.

```python
while True:

    value = int(input("Digite un número entero positivo: "))
    print("Número ingresado: ", value)
    a = isinstance(value, int)
    if a == True and value > 0:
        fact = 1
        for i in range (1, value + 1):
            fact = fact*i
        print(f'El factorial de {value} corresponde a: ', fact)
    else:
        print("Por favor, introduzca un número entero positivo")
```
